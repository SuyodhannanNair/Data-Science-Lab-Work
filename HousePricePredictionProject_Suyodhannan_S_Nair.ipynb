```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 1. Load Dataset
df = pd.read_csv("/content/House Price Prediction Dataset.csv")

# 2. Preprocessing
df = df.drop(columns=['Id'])

# Save original categorical columns
categorical_cols = ['Location', 'Condition', 'Garage']

# One-hot encoding
df_encoded = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Features and Target
X = df_encoded.drop('Price', axis=1)
y = df_encoded['Price']

# 3. Train Model
model = LinearRegression()
model.fit(X, y)

# Save column structure
model_columns = X.columns


# 4. Function to take user input
def get_user_input():
    print("\nEnter House Details:")

    area = float(input("Area (sq ft): "))
    bedrooms = int(input("Number of Bedrooms: "))
    bathrooms = int(input("Number of Bathrooms: "))
    floors = int(input("Number of Floors: "))
    age = int(input("Age of House: "))

    location = input("Location (Urban/Suburban/Rural): ")
    condition = input("Condition (Poor/Fair/Good/Excellent): ")
    garage = input("Garage (Yes/No): ")

    # Create dictionary
    user_data = {
        'Area': area,
        'Bedrooms': bedrooms,
        'Bathrooms': bathrooms,
        'Floors': floors,
        'Age': age,
        'Location': location,
        'Condition': condition,
        'Garage': garage
    }

    return pd.DataFrame([user_data])


# 5. Prepare input for prediction
def preprocess_input(input_df):
    input_encoded = pd.get_dummies(input_df)

    # Align with training columns
    input_encoded = input_encoded.reindex(columns=model_columns, fill_value=0)

    return input_encoded


# 6. Prediction loop
while True:
    user_df = get_user_input()
    processed_input = preprocess_input(user_df)

    prediction = model.predict(processed_input)

    print("\n🏠 Estimated House Price: ₹{:,.2f}".format(prediction[0]))

    cont = input("\nDo you want to predict another house? (yes/no): ")
    if cont.lower() != 'yes':
        print("Exiting...")
        break
```
