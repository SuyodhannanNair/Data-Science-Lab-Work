```python
planets = [
    ['Mercury', 'Venus', 'Earth'],
    ['Mars', 'Jupiter', 'Saturn'],
    ['Uranus', 'Neptune', 'Pluto']
]

result = [planet
          for sublist in planets
          for planet in sublist
          if len(planet) < 6]

print(result)
```
