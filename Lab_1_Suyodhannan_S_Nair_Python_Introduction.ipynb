**Problem 1**

```python
TP = int(input("Enter number of true positives: ")) #TP - True positives, the number of correctly predicted flood cases
FP = int(input("Enter number of false positives: ")) #FP - False positives, the number of non-flood cases incorrectly predicted as flood
TN = int(input("Enter number of true negatives: ")) #"#TN - True negatives, the number of correctly predicted non-flood cases
FN = int(input("Enter number of false negatives: ")) #FN - False negatives, the number of flood cases incorrectly predicted as non-flood cases

totalpred= TP + FP + TN + FN #Sum of predictions
acc = (TP + TN) / (TP + FP + TN + FN) #Calculating accuracy
prec = TP / (TP + FP) #Calculating precision
rec= TP / (TP + FN) #Calculating recall
f1 = (2 * prec * rec) / (prec + rec) #Calculating F1 score

print("The results calculated are:-") #output statement
print("Total Predictions = ", totalpred) #Display the number of total predictions
print("Accuracy = ", acc) #Display accuracy
print("Precision = ", prec) #Display precision
print("Recall = ", rec) #Display recall
print("F1 Score = ", f1) #Display F! score08
10

```

**Problem 2**

```python
SSR = int(input("Enter the value of SSR: ")) #Value of sum of squares due to regression
SSE = int(input("Enter the value of SSE: ")) #Value of sum of squares due to error
SST = SSR + SSE #SST is the total sum of squares obtained by adding SSR and SSE
rsq = SSR / SST #rsq is R²(R squared), obtained by dividing SSR by SST

print("The results are:-") #Output statement
print("SST = ", SST) #Display the value of total sum of squares
print("R² = ", rsq) #Display R² value

if rsq > 0.7:
    print("R² is greater than 0.7, therefore, the model explains more than 70% of variance.")
else:
    print("R² is lesser than 0.7, therefore, the model explains less than or equal to 70% of variance.")
```

**Problem 3**

```python
errors = eval(input("Enter a list of errors: ")) #User creates a list of errors how much ever no of errors it might be, not just 5
n = len(errors)
sumsq = 0
for i in range(n):
    sumsq += errors[i] ** 2 #Calculate the sum of squares of errors
mse = sumsq / n #Calculate Mean Squared Error using the formula
print("MSE = ", mse) #Display mean squared error
```
